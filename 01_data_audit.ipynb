```python
import pandas as pd
import os
```

```python
BASE_PATH = r"C:\ml 2026\dataset"

TRAIN_PATH = os.path.join(BASE_PATH, "train")
TEST_PATH = os.path.join(BASE_PATH, "test")
```

```python
train_s1 = pd.read_csv(
    os.path.join(TRAIN_PATH, "train_source1.tsv"),
    sep="\t"
)

train_s2 = pd.read_csv(
    os.path.join(TRAIN_PATH, "train_source2.tsv"),
    sep="\t"
)

train_s3 = pd.read_csv(
    os.path.join(TRAIN_PATH, "train_source3.tsv"),
    sep="\t"
)

ground_truth = pd.read_csv(
    os.path.join(TRAIN_PATH, "train_ground_truth.tsv"),
    sep="\t"
)
```

```python
test_s1 = pd.read_csv(
    os.path.join(TEST_PATH, "test_source1.tsv"),
    sep="\t"
)

test_s2 = pd.read_csv(
    os.path.join(TEST_PATH, "test_source2.tsv"),
    sep="\t"
)

test_s3 = pd.read_csv(
    os.path.join(TEST_PATH, "test_source3.tsv"),
    sep="\t"
)
```

```python
def count_matches(value):

    if pd.isna(value):
        return 0

    value = str(value).strip()

    if value == "":
        return 0

    return len(value.split(","))
ground_truth["match_count"] = (
    ground_truth["matched_entity_ids"]
    .apply(count_matches)
)
print(
    ground_truth["match_count"]
    .value_counts()
    .sort_index()
)
print(
    "Zero matches:",
    (ground_truth["match_count"] == 0).sum()
)

print(
    "One match:",
    (ground_truth["match_count"] == 1).sum()
)

print(
    "Multiple matches:",
    (ground_truth["match_count"] > 1).sum()
)
```
