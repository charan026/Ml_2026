```python
def normalize_text(text):
    if pd.isna(text):
        return ""

    text = str(text)

    text = unicodedata.normalize("NFKC", text)

    text = text.lower()

    text = text.strip()

    text = re.sub(r"\s+", " ", text)

    return text
```

```python
examples = [
    "  Prime   Money  ",
    "Orelee's Barbershop",
    "B+ Retail Inc",
    "  Christ   Chapel  "
]

for x in examples:
    print("Original :", repr(x))
    print("Normalized:", repr(normalize_text(x)))
    print()
```

```python
def normalize_alnum(text):
    text = normalize_text(text)

    chars = []
    for ch in text:
        if ch.isalnum() or ch.isspace():
            chars.append(ch)
        else:
            chars.append(" ")

    text = "".join(chars)
    text = re.sub(r"\s+", " ", text)

    return text.strip()
```

```python
def normalize_compact(text):
    text = normalize_text(text)

    text = re.sub(r"[^a-z0-9]+", "", text)

    return text
examples = [
    "B+ Retail Inc",
    "Orelee's Barbershop",
    "Delta-Telecommunication Inc."
]

for x in examples:
    print("Original :", x)
    print("Compact  :", normalize_compact(x))
    print()
```

```python
LEGAL_SUFFIXES = [
    "private limited",
    "pvt limited",
    "pvt ltd",
    "private ltd",
    "public limited",
    "public ltd",
    "limited liability company",
    "l l c",
    "llc",
    "incorporated",
    "inc",
    "corporation",
    "corp",
    "company",
    "co",
    "limited",
    "ltd",
    "llp",
    "pvt"
]
```

```python
def normalize_name_no_suffix(text):
    text = normalize_alnum(text)

    if not text:
        return ""

    tokens = text.split()

    while tokens:
        current = " ".join(tokens[-3:])

        if current in LEGAL_SUFFIXES:
            tokens = tokens[:-3]
            continue

        current = " ".join(tokens[-2:])

        if current in LEGAL_SUFFIXES:
            tokens = tokens[:-2]
            continue

        current = tokens[-1]

        if current in LEGAL_SUFFIXES:
            tokens = tokens[:-1]
            continue

        break

    return " ".join(tokens)
```

```python
def normalize_name_ampersand(text):
    text = normalize_text(text)

    text = re.sub(r"\band\b", " & ", text)
    text = re.sub(r"\s*&\s*", " & ", text)
    text = re.sub(r"\s+", " ", text)

    return text.strip()
```

```python
def normalize_address(text):
    if pd.isna(text):
        return ""

    text = normalize_text(text)

    # Replace punctuation with spaces
    chars = []
    for ch in text:
        if ch.isalnum() or ch.isspace():
            chars.append(ch)
        else:
            chars.append(" ")

    text = "".join(chars)
    text = re.sub(r"\s+", " ", text)

    return text.strip()
```

```python
ADDRESS_ABBREVIATIONS = {
    "rd": "road",
    "st": "street",
    "ave": "avenue",
    "av": "avenue",
    "blvd": "boulevard",
    "dr": "drive",
    "ln": "lane",
    "pkwy": "parkway",
    "hwy": "highway",
    "rdg": "ridge",
    "ct": "court",
    "pl": "place",
    "trl": "trail",
    "sq": "square"
}
```

```python
def normalize_address_abbreviations(text):
    text = normalize_address(text)

    if not text:
        return ""

    tokens = text.split()

    tokens = [
        ADDRESS_ABBREVIATIONS.get(token, token)
        for token in tokens
    ]

    return " ".join(tokens)
address_examples = [
    "123 Main Rd",
    "123 Main Road",
    "45 MG St",
    "45 MG Street",
    "10 Park Ave",
    "10 Park Avenue"
]

for x in address_examples:
    print("Original   :", x)
    print("Normalized :", normalize_address_abbreviations(x))
    print()
```

```python
POSTAL_PATTERN = re.compile(r"\b\d{5,6}\b")

def extract_postal_code(text):
    if pd.isna(text):
        return ""

    text = normalize_text(text)

    match = POSTAL_PATTERN.search(text)

    if match:
        return match.group(0)

    return ""
```

```python
def normalize_address_tokens(text):
    text = normalize_address_abbreviations(text)

    if not text:
        return ""

    tokens = text.split()

    tokens = sorted(set(tokens))

    return " ".join(tokens)
```

```python
def add_name_normalizations(df):
    df["name_normalized"] = df["business_name"].apply(normalize_text)
    df["name_alnum"] = df["business_name"].apply(normalize_alnum)
    df["name_compact"] = df["business_name"].apply(normalize_compact)
    df["name_no_suffix"] = df["business_name"].apply(normalize_name_no_suffix)
    df["name_ampersand"] = df["business_name"].apply(normalize_name_ampersand)

    return df
```

```python
def add_address_normalizations(df):
    df["address_normalized"] = df["business_address"].apply(normalize_address)
    df["address_abbrev"] = df["business_address"].apply(normalize_address_abbreviations)
    df["postal_code"] = df["business_address"].apply(extract_postal_code)
    df["address_tokens"] = df["business_address"].apply(normalize_address_tokens)

    return df
```

```python
train_s2 = add_name_normalizations(train_s2)
train_s2 = add_address_normalizations(train_s2)

print("S2 normalization complete")
```

```python
train_s3 = add_name_normalizations(train_s3)
train_s3 = add_address_normalizations(train_s3)

print("S3 normalization complete")
```
